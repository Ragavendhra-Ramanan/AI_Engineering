**Problem Statement**

You are an AI Engineer at a subscription-based streaming service called "StreamVault."
The company has collected data on 10,000 customers over the past year, and they want you to predict which customers will churn (cancel their subscription) in the next 30 days.

However, there is a twist. 
The raw dataset contains subtle patterns that are only visible through proper feature engineering. 
Your task is not just to build a model, but to discover the hidden patterns through systematic feature creation and selection.
